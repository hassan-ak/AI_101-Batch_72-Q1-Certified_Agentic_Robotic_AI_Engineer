```python
def factorial(n):
    # Base case
    if n == 0:
        return 1
    # Recursive case
    else:
        return n * factorial(n-1)

# Example usage
print(factorial(5))
# n = 5  factorial(5) => if n =0 return 1 else return 5 * factorial(5-1)  5 * 24 = 120
# n = 4 factorial(4) => if n =0 return 1 else return 4 * factorial(4-1) 4 * 6 = 24
# n = 3 factorial(3) => if n =0 return 1 else return 3 * factorial(3-1) 3 *2 =6
# n = 2 factorial(2) => if n =0 return 1 else return 2 * factorial(2-1) 2 * 1= 2
# n = 1 factorial(1) => if n =0 return 1 else return 1 * factorial(1-1) # 1 * 1 = 1
# n = 0 factorial(0) => if n =0 return 1 else return 4 * factorial(4-1)
```

```python

try:
    num1 = float(input("Enter the first number: "))
    num2 = float(input("Enter the second number: "))
    result = num1 / num2
except ValueError:
    print("Error: Invalid input. Please enter numbers.")
except ZeroDivisionError:
    print("Error: Cannot divide by zero.")
else:
    print(f"The result is: {result}")
finally:
    print("Thank you for using the program!")
```

```python
def my_division(a, b):
    if b == 1:
        raise ValueError("Division by is not allowed!")  # Raising an exception
    return a / b
```

```python
try:
  my_division(a=5, b=1)
except ValueError as e:
  print(f'{e}')
```
