```python
def modify_value(x):
    x = 10
    print("Within function:", x)
    return x

# Immutable object (integer)
x = 5
print("Original:", x)
modify_value(x)
print("After function:", x)
```

```python
def modify_list(lst):
    lst.append(4)
    print("Within function: ", lst, " - ID:", id(lst))

# Mutable object (list)
lst = [1, 2, 3]
print("Original:", lst, " - ID:", id(lst))
modify_list(lst)
print("After function:", lst, " - ID:", id(lst))
```

```python
def sum_unlimit(*a : int) -> int:
  sum = 0
  for i in a:
    sum = sum + i
  return sum
```

```python
sum_unlimit(2, 3, 45, 67, 45, 34, 23, 23, 46, 78)
```
